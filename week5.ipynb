Load the dataset.



Find the missing values in each column.



Remove the Cabin column.



Replace missing Age values with the median.



Replace missing Embarked values with the mode.



Check whether any missing values remain.

```python

import pandas as pd

# Load the Titanic dataset
df = pd.read_csv('train.csv')

# Display first five rows
print("Original Dataset:")
print(df.head())

# Find missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())

# Drop the Cabin column
df = df.drop('Cabin', axis=1)

# Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with mode
df['Embarked'] = df['Embarked'].fillna(
    df['Embarked'].mode()[0]
)

# Check remaining missing values
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python

import pandas as pd
import numpy as np

# Generate performance scores
np.random.seed(42)

performance_scores = np.random.normal(
    loc=75, scale=10, size=100
)

# Insert some outliers
performance_scores[10] = 150
performance_scores[25] = -20
performance_scores[50] = 5

# Create DataFrame
employee_df = pd.DataFrame({
    'EmployeeID': range(1, 101),
    'PerformanceScore': performance_scores.round(0)
})

# Display first 10 rows
print("Original Data:")
print(employee_df.head(10))

# Calculate Q1 and Q3
Q1 = employee_df['PerformanceScore'].quantile(0.25)
Q3 = employee_df['PerformanceScore'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate upper and lower bounds
upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Upper Bound:", upper_bound)
print("Lower Bound:", lower_bound)

# Identify outliers
outliers = employee_df[
    (employee_df['PerformanceScore'] > upper_bound) |
    (employee_df['PerformanceScore'] < lower_bound)
]

print("\nOutliers:")
print(outliers[['EmployeeID', 'PerformanceScore']])

# Apply capping
employee_df.loc[
    employee_df['PerformanceScore'] > upper_bound,
    'PerformanceScore'
] = upper_bound

employee_df.loc[
    employee_df['PerformanceScore'] < lower_bound,
    'PerformanceScore'
] = lower_bound

# Display scores after capping
print("\nPerformance Scores After Capping:")
print(employee_df['PerformanceScore'])
```

```python

import pandas as pd
import numpy as np

# Generate temperature data
np.random.seed(43)

temp_data = pd.DataFrame({
    'CityName': [
        'London', 'Paris', 'New York', 'Tokyo',
        'Berlin', 'Rome', 'Sydney', 'Cairo'
    ],
    'AvgTempC': np.random.uniform(10, 30, 8).round(1),
    'Humidity': np.random.uniform(50, 90, 8).round(1)
})

print("Temperature Data:")
print(temp_data)

# Create city population data
city_pop_data = pd.DataFrame({
    'City': [
        'London', 'Paris', 'New York', 'Tokyo',
        'Sydney', 'Dubai', 'Beijing'
    ],
    'Country': [
        'UK', 'France', 'USA', 'Japan',
        'Australia', 'UAE', 'China'
    ],
    'PopulationMillions': [9, 2.1, 8.4, 14, 5.3, 3.1, 21.5]
})

print("\nCity Population Data:")
print(city_pop_data)

# Merge both datasets
merged_data = pd.merge(
    temp_data,
    city_pop_data,
    left_on='CityName',
    right_on='City',
    how='inner'
)

print("\nIntegrated Data:")
print(merged_data)
```
