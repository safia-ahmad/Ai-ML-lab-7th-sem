```python

import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Normalization
minmax = MinMaxScaler()
normalized_data = minmax.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=iris.feature_names
)

print("Normalized Data:")
print(normalized_df.head())

# Standardization
standard = StandardScaler()
standardized_data = standard.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=iris.feature_names
)

print("Standardized Data:")
print(standardized_df.head())
```

```python

import pandas as pd

# Column names of Adult dataset
columns = [
    'age', 'workclass', 'fnlwgt', 'education',
    'education_num', 'marital_status', 'occupation',
    'relationship', 'race', 'sex',
    'capital_gain', 'capital_loss',
    'hours_per_week', 'native_country', 'income'
]

# Load dataset
adult_df = pd.read_csv(
    'adult.data',
    names=columns,
    skipinitialspace=True
)

# Define age bins
bins = [0, 30, 45, 60, adult_df['age'].max()]

# Define category labels
labels = [
    'Young Adult',
    'Middle-Aged',
    'Senior Adult',
    'Elderly'
]

# Discretize age
adult_df['age_group'] = pd.cut(
    adult_df['age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Print categories and their counts
print(adult_df['age_group'].value_counts())
```

```python

import pandas as pd

# Load ratings data
ratings_df = pd.read_csv(
    'u.data',
    sep='\t',
    names=['user_id', 'item_id', 'rating', 'timestamp']
)

# Load movie data
movies_df = pd.read_csv(
    'u.item',
    sep='|',
    encoding='latin-1',
    header=None,
    usecols=[0, 1],
    names=['movie_id', 'movie_title']
)

# Calculate average rating and number of ratings
movie_ratings = ratings_df.groupby('item_id').agg(
    avg_rating=('rating', 'mean'),
    num_ratings=('rating', 'count')
).reset_index()

# Merge ratings with movie titles
movie_stats_df = pd.merge(
    movies_df,
    movie_ratings,
    left_on='movie_id',
    right_on='item_id'
)

# Select required columns
movie_stats_df = movie_stats_df[
    ['movie_title', 'avg_rating', 'num_ratings']
]

# Sort by number of ratings
movie_stats_df = movie_stats_df.sort_values(
    by='num_ratings',
    ascending=False
)

# Print top 10 movies
print(movie_stats_df.head(10))
```
