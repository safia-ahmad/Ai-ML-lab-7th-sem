```python

import numpy as np

# AND gate inputs and outputs
X_and = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_and = np.array([0, 0, 0, 1])

# Initialize random weights and bias
weights = np.random.rand(2) * 0.1
bias = np.random.rand(1)[0] * 0.1

# Activation function
def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0

# Learning parameters
learning_rate = 0.1
epochs = 20

# Training the Perceptron
for epoch in range(epochs):

    for x, y_true in zip(X_and, y_and):

        # Calculate weighted sum
        z = np.dot(x, weights) + bias

        # Predict output
        prediction = step_function(z)

        # Calculate error
        error = y_true - prediction

        # Update weights
        weights = weights + learning_rate * error * x

        # Update bias
        bias = bias + learning_rate * error

# Test the trained Perceptron
print("AND Gate Predictions:")

for x in X_and:
    z = np.dot(x, weights) + bias
    prediction = step_function(z)
    print(x, ":", prediction)
```

```python

import numpy as np
import pandas as pd

from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Sample data for shapes
# Features: NumCorners, AspectRatio
# Target: 0 = Square, 1 = Circle

X_shapes = np.array([
    [4, 1.0], [4, 1.05], [4, 0.95],  # Squares
    [0, 1.0], [0, 1.1], [0, 0.9],    # Circles
    [4, 0.5], [0, 2.0]               # Noisy points
])

# Corresponding labels
y_shapes = np.array([0, 0, 0, 1, 1, 1, 0, 1])

# Split data into training and testing sets
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_shapes, y_shapes, test_size=0.3, random_state=42
)

# Display training data
print("Shape Classification Data:\n", X_train_s)
print("\nShape Target:\n", y_train_s)

# Initialize MLP Classifier
model = MLPClassifier(
    hidden_layer_sizes=(5,),
    max_iter=500,
    random_state=42,
    solver='adam'
)

# Train the model
model.fit(X_train_s, y_train_s)

# Make predictions
y_pred = model.predict(X_test_s)

# Calculate accuracy
accuracy = accuracy_score(y_test_s, y_pred)

print("\nPredictions:", y_pred)
print("Actual Values:", y_test_s)
print("Accuracy:", accuracy)
```
