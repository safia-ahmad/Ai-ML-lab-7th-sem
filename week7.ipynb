```python

# Import necessary libraries
import pandas as pd
import numpy as np

# Import function to divide data into training and testing sets
from sklearn.model_selection import train_test_split

# Import KNN classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Import accuracy metric
from sklearn.metrics import accuracy_score


# STEP 1: Create the dataset
data = {
    'Size_cm': [7, 7.5, 6.8, 18, 19, 17.5, 8.5, 9, 8.2, 7.2],

    'Color_Intensity': [0.8, 0.75, 0.82, 0.6, 0.58, 0.62, 0.9, 0.88, 0.92, 0.78],

    'FruitType': ['Apple', 'Apple', 'Apple', 'Banana', 'Banana',
                  'Banana', 'Orange', 'Orange', 'Orange', 'Apple']
}

# Convert dictionary into a Pandas DataFrame
df = pd.DataFrame(data)


# STEP 2: Separate input features (X) and target (y)

# X contains the characteristics used to identify fruits
X = df[['Size_cm', 'Color_Intensity']]

# y contains the actual fruit names
y = df['FruitType']


# STEP 3: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Display training data
print("Training Data:\n", X_train.head())
print("\nFruit Types:\n", y_train.head())


# STEP 4: Create the KNN model with 3 nearest neighbours
model = KNeighborsClassifier(n_neighbors=3)

# STEP 5: Train the model
model.fit(X_train, y_train)

# STEP 6: Predict fruit types using test data
y_pred = model.predict(X_test)

# Display predictions
print("\nPredicted Fruit Types:", y_pred)

# STEP 7: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python

# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

# Import train_test_split
from sklearn.model_selection import train_test_split

# Import Linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Import Mean Squared Error metric
from sklearn.metrics import mean_squared_error


# STEP 1: Create the dataset

# Input feature: Temperature in Celsius
temp = np.array([10, 15, 20, 25, 30, 12, 18, 23, 28, 32]).reshape(-1, 1)

# Output/target: Electricity consumption in kWh
kwh = np.array([50, 60, 75, 90, 105, 55, 70, 85, 95, 110])


# STEP 2: Split the data into training and testing sets

X_train_elec, X_test_elec, y_train_elec, y_test_elec = train_test_split(
    temp, kwh, test_size=0.3, random_state=42
)

# Display training data
print("Training Temperature Data:\n", X_train_elec[:5])
print("\nTraining Electricity Data:\n", y_train_elec[:5])


# STEP 3: Create the Linear Regression model
model = LinearRegression()

# STEP 4: Train the model
model.fit(X_train_elec, y_train_elec)


# STEP 5: Make predictions on test data
y_pred = model.predict(X_test_elec)

print("\nPredicted Electricity Consumption:\n", y_pred)


# STEP 6: Calculate Mean Squared Error
mse = mean_squared_error(y_test_elec, y_pred)

print("\nMean Squared Error:", mse)


# STEP 7: Visualize the data and regression line

# Scatter plot of original training data
plt.scatter(X_train_elec, y_train_elec, color='blue',
            label='Actual Training Data')

# Draw the regression line
plt.plot(X_train_elec, model.predict(X_train_elec),
         color='red', label='Regression Line')

# Add graph labels and title
plt.xlabel("Temperature (°C)")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Temperature vs Electricity Consumption")

# Display legend and graph
plt.legend()
plt.show()
```
