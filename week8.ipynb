```python

# Import necessary libraries
import pandas as pd
import numpy as np

# Import Decision Tree algorithm
from sklearn.tree import DecisionTreeClassifier

# Import train_test_split
from sklearn.model_selection import train_test_split

# Import accuracy metric
from sklearn.metrics import accuracy_score


# STEP 1: Create the dataset

data = {
    # Generate 100 random ages between 20 and 60
    'Age': np.random.randint(20, 60, 100),

    # Generate purchase frequency between 0.5 and 5
    'PurchaseFrequency': np.random.uniform(0.5, 5, 100).round(1),

    # Generate responses: 0 = No, 1 = Yes
    'RespondedToCampaign': np.random.choice(
        [0, 1], 100, p=[0.6, 0.4]
    )
}


# STEP 2: Make response somewhat dependent on age and frequency

for i in range(100):

    # Younger customers with higher purchase frequency
    # are more likely to respond
    if data['Age'][i] < 35 and data['PurchaseFrequency'][i] > 3:
        data['RespondedToCampaign'][i] = 1

    # Older customers with low purchase frequency
    # are less likely to respond
    elif data['Age'][i] > 50 and data['PurchaseFrequency'][i] < 1.5:
        data['RespondedToCampaign'][i] = 0


# STEP 3: Convert dictionary into DataFrame
df = pd.DataFrame(data)


# STEP 4: Separate input features (X) and target (y)

X = df[['Age', 'PurchaseFrequency']]
y = df['RespondedToCampaign']


# STEP 5: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)


# Display training data
print("Campaign Response Data:\n", X_train.head())

print("\nCampaign Response Target:\n", y_train.head())


# STEP 6: Create Decision Tree model
# max_depth=3 limits the tree to 3 levels

model = DecisionTreeClassifier(max_depth=3)


# STEP 7: Train the model
model.fit(X_train, y_train)


# STEP 8: Make predictions
y_pred = model.predict(X_test)

print("\nPredicted Responses:\n", y_pred)


# STEP 9: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import Logistic Regression algorithm
from sklearn.linear_model import LogisticRegression

# Import train_test_split
from sklearn.model_selection import train_test_split

# Import accuracy metric
from sklearn.metrics import accuracy_score


# STEP 1: Set random seed
np.random.seed(42)


# STEP 2: Create the dataset

# Generate 100 random numbers of suspicious keywords (0 to 9)
keywords = np.random.randint(0, 10, 100)

# Calculate spam probability using sigmoid function
spam_prob = 1 / (1 + np.exp(-(keywords - 5)))

# Generate spam labels: 0 = Not Spam, 1 = Spam
is_spam = (np.random.rand(100) < spam_prob).astype(int)


# STEP 3: Create a DataFrame
df = pd.DataFrame({
    'NumSuspiciousKeywords': keywords,
    'IsSpam': is_spam
})


# STEP 4: Separate input features (X) and target (y)

X = df[['NumSuspiciousKeywords']]  # Input feature
y = df['IsSpam']                   # Output/target


# STEP 5: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)


# Display training data
print("Spam Detection Data:\n", X_train.head())

print("\nSpam Target:\n", y_train.head())


# STEP 6: Create Logistic Regression model
model = LogisticRegression()


# STEP 7: Train the model
model.fit(X_train, y_train)


# STEP 8: Make predictions
y_pred = model.predict(X_test)

print("\nPredicted Spam Labels:\n", y_pred)


# STEP 9: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```
